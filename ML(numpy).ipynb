assignment,view  and copy

```python
a=np.array([1,2])
print(a)
b=a
print(np.shares_memory(a,b))
print(a is b)
b[0]=10
print(b)
print(a)
print(b.base)
print(b.base is a)
print(a.ctypes.data)
print(b.ctypes.data)
```

```python
B=a.view()
print(B)
B[1]=20
print(B)
print(a)
print(np.shares_memory(a,B))
print(a is B)
print(B.base)
print(B.base is a)
print(a.ctypes.data)
print(B.ctypes.data)
```

```python
c=a.copy()
print(c)
c[1]=100
print(c)
print(a)
print(np.shares_memory(a,c))
print(a is c)
print(c.base)
print(c.base is a)
print(a.ctypes.data)
print(c.ctypes.data)
```
